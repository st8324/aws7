```python
from common import ask
from models.train_price_model import load, predict_price as ml_predict

model, scaler = load()   # 앞에서 저장한 시세 모델

def predict_price(car_type: str, year_age: int, mileage_km: int, accident: bool) -> dict:
    """중고차 예상 시세를 만 원 단위로 알려준다.

    Args:
        car_type: 차종. 경차, 세단, SUV 중 하나
        year_age: 연식(몇 년 된 차인지). 올해는 2026년
        mileage_km: 주행거리(km)
        accident: 사고 이력이 있으면 True
    """
    print(f"  >> 함수 호출: {car_type}, {year_age}년, {mileage_km}km, 사고={accident}")
    price = ml_predict(model, scaler, car_type, year_age, mileage_km, accident)
    return {"price_manwon": round(price)}

r = ask("21년식 SUV에 4만 km 타고 무사고면 얼마쯤 하나요?",
        tools=[predict_price],
        system_instruction="너는 중고차 상담사다. 가격은 반드시 predict_price 결과만 말한다.")
print(r.text)
```
