```python
with open("memo.txt", "w", encoding="utf-8") as f:      # 덮어쓰기
    f.write("첫째 줄\n")

with open("memo.txt", "a", encoding="utf-8") as f:      # 이어 쓰기
    f.write("둘째 줄\n")
    
with open("memo.txt", "r", encoding="utf-8") as f:
    text = f.read()
    print(text)
print("첫번째 읽기 완료")
with open("memo.txt", "r", encoding="utf-8") as f:
    for line in f:
        print(line.strip())
```

### JSON -> dict, dict -> JSON

```python
import json

dataDict = {"name": "홍길동", "age": 25, "skills": ["python", "java"], "car": None}

s = json.dumps(dataDict, ensure_ascii=False)       # dict → 문자열

print(type(s), s)
print(type(dataDict), dataDict)
```

```python
s = '{"name" : "홍길동", "age" : 21}'
loaded = json.loads(s)                         # 문자열 → dict

print(type(loaded), loaded)
```

```python
with open("user.json", "w", encoding="utf-8") as f:
    json.dump(s, f, ensure_ascii=False, indent=2)

with open("user.json", "r", encoding="utf-8") as f:
    user = json.load(f) # load(파일) : 파일에서 읽어옴
    print(type(user), user)
    userDict = json.loads(user) # loads(str) : 문자열을 읽어와 dict로 변환
    print(type(userDict), userDict)
```
