```python
fruits = ["사과", "바나나", "포도", "딸기", "수박"]
prices = [3000, 1500, 5000, 6000, 12000]

# 위 두 리스트를 이용하여 dataFrame으로 변환
import pandas as pd
df = pd.DataFrame({"fruit" : fruits, "price": prices})
```

```python
# 가격이 5천원 이상이 과일을 조회
print(df[df["price"] >= 5000])
# 각 과일의 10프로 할일된 가격을 조회
df["discount"] = df["price"] * 0.9
print(df)
```
