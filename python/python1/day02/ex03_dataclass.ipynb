```python
from dataclasses import dataclass, field, asdict

@dataclass
class Book:
    title: str # 원래는 클래스 변수이어야 하지만 @dataclass 때문에 인스턴스 변수로 생성
    price: int
    author: str = "미상"
    tags: list[str] = field(default_factory=list)

    def discounted(self, rate: float) -> int:
        return int(self.price * (1 - rate))

    # 생성자를 자동으로 추가(모든 인스턴스 변수를 매개변수로 갖는)

    # 두 객체를 비교하는 기능을 추가


b1 = Book("파이썬 입문", 20000)
b2 = Book("파이썬 입문", 20000)
print(b1)            # Book(title='파이썬 입문', price=20000, author='미상', tags=[])
# 원래 ==는 주소를 비교. @dataclass가 필드의 값이 같은지를 비교하도록 해줌
print(b1 == b2)      # True
print(asdict(b1))    # dict로 변환
```
