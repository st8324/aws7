```python
# cafe_sales.csv 파일을 읽어와서 데이터프레임에 가져오는 코드를 작성하세요
import pandas as pd

df = pd.read_csv("cafe_sales.csv")
print(df)
```

```python
# menu, price의 두열만 선택 한후 앞 3행을 출력하는 코드를 작성하세요.
df[["menu", "price"]].head(3)
```

```python
# price의 평균과 최대값을 구하는 코드를 작성하세요.
print(df["price"].mean(), df["price"].max())
```

```python
# quantity의 합계(전체 판매 수량)를 구하는 코드를 작성하세요.
print(df["quantity"].sum())
```

```python
# 매출(sales) 컴럼을 추가
# 가격 X 수량
df["sales"] = df["price"] * df["quantity"]
df.head(5)
```

```python
# 강남점 주문만 주문 건수와 매출 합계를 구함
gangnam = df[df["store"] == "강남점"]
gangnam.head(5)
print(len(gangnam)) # 주문건수
print(gangnam["sales"].sum()) # 매출 합계
```

```python
# 디저트 중 매출(sales)이 20000 이산인 주문을 출력
print(df[(df["sales"] >= 20000) & (df["category"] == "디저트")])
```

```python
# 지점별 매출(sales)를 조회
print(df.groupby("store")["sales"].sum())
```

```python
# 매출이 추가된 dataframe을 cafe_sales2.csv로 저장
df.to_csv("cafe_sales2.csv")
```
