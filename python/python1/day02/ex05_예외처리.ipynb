# 예외 발생 시키기
* 코드로는 예외가 발생되지 않지만 상황에 의해 예외를 발생 시켜야 하는 경우 사용
* `raise`를 이용

```python
class InsufficientBalanceError(Exception): # 사용자 정의 예외 클래스
    pass

def withdraw(balance: int, amount: int) -> int: # 출금함수. 예외가 발생할 수 있음
    if amount <= 0:
        raise ValueError("출금액은 0보다 커야 합니다") # 예외 발생 시키키
    if amount > balance:
        raise InsufficientBalanceError(f"잔액 부족: {balance}원") # 예외 발생 시키키
    return balance - amount

for amount in [300, -1, 5000]:
    try:
        print("남은 잔액:", withdraw(1000, amount))
    except (ValueError, InsufficientBalanceError) as e:
        print("실패:", e)
```

```python
# 예외처리가 된 함수
def divide(a: str, b: str) -> float | None:
    try:
        return int(a) / int(b)
    except ValueError:
        print("숫자를 입력하세요")
    except ZeroDivisionError:
        print("0으로 나눌 수 없습니다")
    except Exception as e:
        print(f"알 수 없는 오류: {type(e).__name__}: {e}")
    return None
```

```python
print(divide("1", "2"))
```

```python
print(divide("1", "a"))
```

```python
print(divide("1", "0"))
```

```python
print(divide("1", None))
```

* else : try문이 정상 종료되면 실행하는 곳. except가 실행되면 else는 실행되지 않음

```python
def parse_age(text: str) -> int:
    try:
        age = int(text)
    except ValueError as e:
        print(f"숫자가 아닙니다: {e}")
        return -1
    else:
        print("변환 성공")
        return age
    finally:
        print("검사 끝")

print(parse_age("25"))
print(parse_age("abc"))
```
