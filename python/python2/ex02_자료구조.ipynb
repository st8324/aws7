## 자료구조
* 데이터를 관리하는 방법
### list
* 중복을 허용
* 순서를 보장
* 수정 가능
* `[]`를 이용
### tuple
* 수정할 수 없는 `list`
* 중복을 허용
* 순서를 보장
* 수정 불가능
* `()`를 이용
### dict
* `key`와 `value`로 구성
* `key`는 중복이 불가능
* 순서를 보장
* 수정 가능
* `{ "kye명" : "값"}`로 이용
### set
* 중복 불가능
* 순서 보장 못함
* 수정 가능
* `{값, 값}`을 이용

````python
# 학생의 정보를 관리하려고 한다
# name은 홍길동이고, age는 17살이다
# 이 때 필요한 student를 만들어서 출력해보세요.
student = { "name" : "홍길동", "age": 17}
print(student)
print(f"이름은 {student["name"]}입니다.") # []는 해당 key가 없으면 에러 발생
print(f"나이는 {student.get("age")}살입니다.") # get()은 해당 key가 없으면 None
````

## 함수
### 함수 정의
* `def` 키워드 이용
* 파이썬은 함수 오버로딩을 제공하지 않음
    * 오버로딩 : 동일한 이름의 함수가 여러 개인 경우
* 키워드 인자를 사용하면 함수 호출할 때 인자 순서를 지키지 않아도 됨
```
def 함수명(변수1, 변수2, ...):
    구현
```
```
def 함수명(변수1:타입힌트, 변수2:타입힌트, ...)->리턴타입힌트:
    구현
```

````python
# Hello를 출력하는 함수
def hello():
    print("Hello")
hello()
````

````python
# 인사말(greet)과 이름(name)이 주어지면 "인사말 ~~~님 반갑습니다. "형태로 출력하는 함수를 
# 선언하고 호출해보세요.
# 추가사항 : 기본 인사말이 안녕하세요.가 되도록 작업
greet = "안녕하세요."
name = "홍길동"

def greeting(name, greet="안녕하세요."):
    print(f"{greet} {name}님 반갑습니다.")

greeting(name, greet)
greeting(name, "하이.")
greeting(name)
greeting(greet="헬로우.", name=name) # 키워드 인자. 키워드 인자를 사용하면 순서가 상관 없음
````

````python
# 리스트에 기본값을 사용하는 함수(잘못된 방식)
def printList(list=[]):
    for item in list:
        print(item)
    list.append(1) # list의 기본값을 []로 초기화하면 어떤 현상이 생기는지 보여주기 위한 코드

printList([1, 2, 3])
printList() # list=[]를 이용
printList() # list=[1]
printList() # list=[1, 1]
````

````python
# 리스트에 기본값을 사용하는 함수(올바른 방식)
def printList(list=None):
    if list is None:
        list = []
    for item in list:
        print(item)
    list.append(1) # list의 기본값을 []로 초기화하면 어떤 현상이 생기는지 보여주기 위한 코드

printList([1, 2, 3])
printList() 
printList() 
printList() 
````

### 함수
* 리턴값은 여러개를 반환할 수 있음
* 반환값은 튜플로 받아옴

````python
# 리스트에서 최소값과 최대값을 반환하는 함수
def min_max(list):
    return min(list), max(list)

min_num, max_num = min_max([1,2,3,4,5]) # 리턴값이 2개이기 때문에 변수 2개로 받아와야 함
print(f"최소값 : {min_num}, 최대값 : {max_num}")
_,max_num = min_max([1,2,3,4,5]) # 리턴값이 2개이기 때문에 변수 2개로 받아와야 함
print(f"최대값 : {max_num}")
````

## 컴프리헨션
* 반복문을 이용하여 list, dict, set, tuple를 코드 한줄을 통해 만드는 방법
```
변수명 = [ 변수 for 변수 in 반복대상 if 조건]
```

````python
# 1부터 10이하의 짝수를 가지고 있는 리스트를 반복문을 작성해보세요.
# list.append(값)
list = []
for i in range(1, 11):
    if i % 2 == 0:
        list.append(i)

print(list)
````

````python
list2 = [ i for i in range(1,11) if i % 2 == 0]
print(list2)
````

````python
# 다음과 같은 dict 있을 때 이름 : 성적 형태로 출력하는 코드를 작성하세요.
students = [
    {"name" : "홍길동", "score" : 85},
    {"name" : "고길동", "score" : 55},
    {"name" : "김둘리", "score" : 75},
    {"name" : "하니", "score" : 100},
]
for student in students:
    print(f"{student["name"]} : {student["score"]}")
````

````python
# 성적이 80점 이상인 학생 이름을 출력하는 코드를 작성하세요.
for student in students:
    if student["score"] >= 80:
        print(student["name"])
# 성적이 80점 이상인 학생 이름을 리스트에 담아 출력하는 코드를 작성하세요.
list2 = []
for student in students:
    if student["score"] >= 80:
        list2.append(student["name"])
print(list2)

list3 = [student["name"] for student in students if student["score"] >= 80]
print(list3)
````
