### 연산자
* 산술 연산자
    * +, -, *, %
    * / : 결과가 실수
    * //: 나눈 결과에서 몫(정수부분)을 가져옴. 자바에서 정수/정수한 결과
    * `a ** b` : a의 b제곱. 자바에서 Math.pow(a,b)
* 논리 연산자
    * &&, ||, ! => and, or, not
* 비교 연산자
    * 연속 비교가 가능
    * 90 <= A <= 100
* 증감 연산자
    * 파이썬에선 없음
* 멤버십 연산자
    * 리스트나 문자열에 특정 값이 있는지 확인할 때 사용
    * `in`, `not in`

````python
# score에 90을 저장
score = 90
# score가 A학점인지 아닌지를 출력하는 코드를 작성하세요. A는 90이상 100이하
print(f"{score}점은 A학점인가요? {90 <= score <= 100}")
print(f"{score}점은 A학점인가요? {90 <= score and score <= 100}")
# name에 홍길동을 저장
name = "홍길동"
# name에 "이"가 들어가는 확인하는 코드를 작성하세요.
print(f"{name}에는 \"이\"가 들어가나요? {"이" in name}")
````

## 조건문
### if문
```
if 조건식:
    실행문
elif 조건식:
    실행문
else:
    실행문
```
### match case문
* break문이 없음
```
match 변수:
    case 값1:
        실행문:
    case 값2:
        실행문:
    case _: # 자바의 default
        실행문
```

````python
# score에 65를 저장하고, score가 60이상이면 Pass, 아니면 Fail이 출력되도록 코드를 작성하세요.
score = 65
if score >= 60:
    print("Pass")
else:
    print("Fail")
````

````python
# month에 1을 저장
month = 1
# month가 12,1,2이면 겨울, month가 3,4,5이면 봄, 6,7,8이면 여름, 9,10,11이면 가을, 아니면 잘못된 월입니다라고 출력하는
# 코드를 작성하세요
match month:
    case 12 | 1 | 2:
        print("겨울")
    case 3 | 4 | 5: 
        print("봄")
    case 6 | 7 | 8:
        print("여름")
    case 9 | 10 | 11:
        print("가을")
    case _:
        print("잘못된 월입니다.")
# (12, 1, 2) : 튜플
if month in (12, 1, 2):
    print("겨울")
elif month in (3, 4, 5):
    print("봄")
elif month in (6,7,8):
    print("여름")
elif month in(9,10,11):
    print("가을")
else:
    print("잘못된 월입니다.")
````

## 반복문
### for문
```
for 변수 in range(시작숫자, 끝숫자+1, 간격):
    실행문
```
```
for 변수 in 리스트: # 리스트 대신 튜플, dict 가능
    실행문
```

````python
# "Hello"를 5번 출력하는 예제
for i in range(5): # 0부터 4까지
    print("Hello", i)
print("--------")
for i in range(1,6): #1부터 5까지
    print("Hello", i)
print("--------")
for i in range(1, 6, 1): # 1부터 5까지 1씩 증가
    print("Hello", i)
````

````python
# 구구단 2단을 출력하는 코드를 작성하세요.
# 2 X 1 = 2
num = 2
for i in range(1, 10):
    print(f"{num} X {i} = {num*i}")
````
