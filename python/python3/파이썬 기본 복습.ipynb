# 조건문

```python
# if문
# 숫자가 음수, 양수, 0인지를 판별하는 예제
# else if => elif
# {} => : 줄바꿈 후 들여쓰기
num = 10
if num > 0 :
    print(f"{num}는 양수")
elif num < 0 :
    print(f"{num}는 양수")
else :
    print(0)
```

```python
# match문(자바의 switch문 역할을 함)
# match문은 break를 안씀
# month가 12, 1, 2이면 겨울, 3, 4, 5이면 봄, 6, 7 , 8이면 여름, 9,10,11이면 가을로
# 판정하는 예제
month = 13
match month:
    case 12 | 1 | 2:
        print(f"{month}는 겨울")
    case 3 | 4 | 5:
        print(f"{month}는 봄")
    case 6 | 7 | 8:
        print(f"{month}는 여름")
    case 9 | 10| 11:
        print(f"{month}는 가을")
    case _: # 자바에서 default
        print(f"{month}는 잘못된 월입니다")
```

```python
# 성적(0~100)이 커트라인(예:60) 이상이면 합격, 아니면 불합격을 출력하는 코드를 작성하세요.
# 단, 함수 이용. 
# 커트라인이 없으면 60(어려울수 있으니 적용 안해도 됨. )

score = 59
cutline = 60

def check_pass(score, cutline=60): # 함수에 기본값을 적용
    if score >= cutline:
        return "합격"
    else:
        return "불합격"

print(f"{score}점은 {check_pass(score)}") #함수 호출할 때 기본값이 있는 매개변수는 생략해도 됨
```

```python
# 이름(name), 나이(age), 주소(address)를 주면 딕셔너리로 만들어 주는 함수를 만들고,
# 다음처럼 호출했을 때 실행되도록 작업

def to_dict(age, name="익명", address="없음"):
    return {
        "name" : name,
        "age" : age,
        "address" : address
    }

user1 = to_dict(21, "홍길동", "서울시")
print(user1)

user2 = to_dict(21, "홍길동") # 홍길동, 21, 없음
print(user2)
user3 = to_dict(21) # 익명, 21, 없음
print(user3)
# user4는 나이 : 21, 주소 : 서울시가 되도록 함수를 호출하세요
user4 = to_dict(21, address="서울시") # 매개변수 2개, 키워드 인자 사용
print(user4)
user5 = to_dict(21, "서울시") # 매개변수 2개, 키워드 인자 미사용
print(user5)
```

```python
# 사람들 정보를 주면 사람들 수와 평균 나이를 반환하는 함수를 만드세요.
users = [
    {"name" : "홍길동", "age" : 21},
    {"name" : "임꺽정", "age" : 41},
    {"name" : "하니", "age" : 15}
]

# 함수를 호출해서 
# 회원 3명. 평균 나이 25.7살
def get_count_and_mean_age(users):
    count = len(users)
    sum_age = 0
    for user in users:
        sum_age += user.get("age")
    mean_age = sum_age / count
    return count, mean_age

count, mean_age = get_count_and_mean_age(users)
print(f"회원 {count}명. 평균 나이 {mean_age:.1f}살")
```

# 컴프리핸션
* 반복문을 이용하여 리스트, 튜플, 셋, 딕셔너리를 만들 때 사용
* 한 줄로 처리

```python
# 앞에서 한 예제를 컴프리핸션을 적용하기 위해 변형
def get_count_and_mean_age2(users):
    count = len(users)

    # 나이들만 모아서 리스트
    ages = []
    for user in users:
       ages.append(user.get("age"))
    
    sum_age = sum(ages)
    mean_age = sum_age / count
    return count, mean_age

count, mean_age = get_count_and_mean_age2(users)
print(f"회원 {count}명. 평균 나이 {mean_age:.1f}살")
```
