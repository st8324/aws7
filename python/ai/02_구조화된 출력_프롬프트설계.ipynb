```python
SYSTEM = """너는 중고차 판매글 작성 도우미다.
규칙:
- 3문장 이내로 쓴다.
- 연식, 주행거리, 사고 여부를 반드시 넣는다.
- '최고', '완벽' 같은 과장 표현은 쓰지 않는다."""
```

```python
car = "차종: SUV, 연식: 2021년, 주행거리: 32000km, 사고"

# 같은 정보, temperature만 바꿔서 비교
for t in [0.0, 1.0]:
    r = ask(car, system_instruction=SYSTEM, temperature=t)
    print(f"[temperature={t}]", r.text)
```

```python
prompt = """중고차 후기를 긍정/부정/중립 중 하나로만 답하라.

후기: 상태가 설명이랑 똑같아요 → 긍정
후기: 사진이랑 너무 달라서 실망 → 부정
후기: 가격은 시세 수준이었어요 → 중립
후기: 딜러분이 친절하고 차도 깨끗했어요 →"""

print(ask(prompt, temperature=0).text)   # 긍정
```

```python
from typing import Literal, Optional
from pydantic import BaseModel
from common import ask

class CarInfo(BaseModel):
    # 인스턴스 변수
    car_type: Optional[Literal["경차", "세단", "SUV", "기타"]]
    year_age: Optional[int]      # 연식(몇 년 된 차인지)
    mileage_km: Optional[int]
    accident: Optional[bool]
    enginecc: Optional[int]
    
# 다음 문장으로 중고차 정보를 확인
text = "19년식 투싼 팔아요. 6만 2천 탔고 사고는 없습니다. 내부 깨끗해요!"

r = ask(
    f"올해는 2026년이다. 다음 판매글에서 차량 정보를 뽑아라. 글에 없는 값은 null로 둔다.\n{text}",
    response_mime_type="application/json",
    response_schema=CarInfo,
    temperature=0,
)
info = r.parsed
```

```python
print(info)
print(info.car_type)
```
